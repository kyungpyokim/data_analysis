## 📊 Pandas 조회 연습 문제 정답 확인

### Q1. `is_active` 열만 추출

* **loc:** `df.loc[:, "is_active"]`
* **iloc:** `df.iloc[:, -2]` 또는 `df.iloc[:, 5]`
> **Tip:** `is_active`는 뒤에서 두 번째 열이므로 `-2`를 사용하면 편리합니다.



### Q2. `age`와 `group` 열만 추출

* **loc:** `df.loc[:, ["age", "group"]]`
* **iloc:** `df.iloc[:, [2, -1]]`
> **Point:** 여러 열을 선택할 때는 리스트 `[]`로 감싸주어야 합니다.



### Q3. `customer_id`가 105인 행(인덱스 4) 추출

* **loc:** `df.loc[4, :]`
* **iloc:** `df.iloc[4, :]` 또는 `df.iloc[-1, :]`

### Q4. `customer_id` 103, 104, 105 행 추출

* **loc:** `df.loc[[2, 3, 4], :]`
* **iloc:** `df.iloc[2:, :]` 또는 `df.iloc[-3:, :]`
> **Difference:** `loc` 슬라이싱(`2:4`)은 끝점인 4를 **포함**하지만, `iloc` 슬라이싱(`2:5`)은 끝점 앞까지인 점을 주의하세요.



### Q5. 'Charlie'의 `group` 정보 추출 (인덱스 2)

* **loc:** `df.loc[2, "group"]`
* **iloc:** `df.iloc[2, -1]`

### Q6. `customer_id` 102~103의 `name`, `age` 추출

* **loc:** `df.loc[[1, 2], ["name", "age"]]`
* **iloc:** `df.iloc[1:3, 1:3]`

### Q7. 'Alice'의 `customer_id`, `name`, `age` 추출 (인덱스 0)

* **loc:** `df.loc[0, ["customer_id", "name", "age"]]`
* **iloc:** `df.iloc[0, :3]`

### Q8. 'Bob'과 'Eva'의 `signup_date`, `is_active` 추출 (인덱스 1, 4)

* **loc:** `df.loc[[1, 4], ["signup_date", "is_active"]]`
* **iloc:** `df.iloc[[1, 4], [3, 5]]` 또는 `df.iloc[[1, 4], [3, -2]]`

---

### 💡 주요 개념 요약

* **`loc`**: 라벨(이름) 기반 조회. 슬라이싱 시 **끝점 포함**.
* **`iloc`**: 정수 위치 기반 조회. 슬라이싱 시 **끝점 미포함**.
* **리스트 활용**: 여러 행/열을 선택할 때는 `df.loc[[행1, 행2], [열1, 열2]]` 형태를 사용합니다.
